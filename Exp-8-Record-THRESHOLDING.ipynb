```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('GTA.jpeg')
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
plt.figure(figsize=(12, 8))
```

```python
plt.subplot(1, 1, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('on')
```

```python
plt.subplot(1, 1, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Grayscale Image')
plt.axis('on')
```

```python
ret, binary = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)
```

```python
plt.subplot(1, 1, 1)
plt.imshow(binary, cmap='gray')
plt.title('Binary Thresholding')
plt.axis('on')
```

```python
ret, binary_inv = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY_INV)
```

```python
plt.subplot(1, 1, 1)
plt.imshow(binary_inv, cmap='gray')
plt.title('Binary Inverse Thresholding')
plt.axis('on')
```

```python
ret, trunc = cv2.threshold(gray_image, 127, 255, cv2.THRESH_TRUNC)
```

```python
plt.subplot(1, 1, 1)
plt.imshow(trunc, cmap='gray')
plt.title('Truncated Thresholding')
plt.axis('on')
```

```python
ret, tozero = cv2.threshold(gray_image, 127, 255, cv2.THRESH_TOZERO)
```

```python
plt.subplot(1, 1, 1)
plt.imshow(tozero, cmap='gray')
plt.title('To Zero Thresholding')
plt.axis('on')
```

```python
plt.tight_layout()
plt.show()
```
